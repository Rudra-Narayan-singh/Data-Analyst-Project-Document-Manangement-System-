### 📁 Simple Document Management System (DMS)

We will build a simple system using Python. It uses a **Class** called `Document` to represent each file, and another class called `DocumentManager` to handle actions like adding, searching, and showing documents.

```
class Document:
    def __init__(self, doc_id, title, content):
        self.doc_id = doc_id
        self.title = title
        self.content = content

    def __str__(self):
        return f"[ID: {self.doc_id}] Title: {self.title}"

class DocumentManager:
    def __init__(self):
        # We store our documents in a list
        self.documents = []

    def add_document(self, doc_id, title, content):
        # Create a new Document object and add it to our list
        new_doc = Document(doc_id, title, content)
        self.documents.append(new_doc)
        print(f"Success: '{title}' added successfully!")

    def list_all_documents(self):
        if not self.documents:
            print("No documents found in the system.")
            return
        print("\n--- Document List ---")
        for doc in self.documents:
            print(doc)

    def search_document(self, keyword):
        print(f"\n--- Searching for: '{keyword}' ---")
        found = False
        for doc in self.documents:
            # Search both title and content (case-insensitive)
            if keyword.lower() in doc.title.lower() or keyword.lower() in doc.content.lower():
                print(f"Found Match -> ID: {doc.doc_id} | Title: {doc.title}")
                print(f"Content Preview: {doc.content[:100]}...")
                found = True
        if not found:
            print("No matching documents found.")
```

### 🚀 Testing our System
Let's add some sample documents and try searching through them!

```
# Initialize our document manager
dms = DocumentManager()

# 1. Add some documents
dms.add_document(101, "Python Basics", "Python is an easy-to-learn programming language.")
dms.add_document(102, "Shopping List", "Buy milk, eggs, bread, and some coffee.")
dms.add_document(103, "Project Plan", "The final deadline for the Python project is next Friday.")

# 2. List all saved documents
dms.list_all_documents()

# 3. Search for a specific word
dms.search_document("python")
dms.search_document("milk")
```
